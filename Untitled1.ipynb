```python
# tf_flowers_cnn.py
"""
Run:
  pip install tensorflow tensorflow-datasets matplotlib
  python tf_flowers_cnn.py

Notes:
- This will download the tf_flowers dataset (~100-200MB).
- If you have limited RAM/GPU, reduce batch_size or image_size.
"""

import tensorflow as tf
import tensorflow_datasets as tfds
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import os

# Config
BATCH_SIZE = 32
IMG_SIZE = 160  # you can use 128/160/224 depending on memory
AUTOTUNE = tf.data.AUTOTUNE
EPOCHS = 12
MODEL_DIR = 'tf_flowers_cnn'

# 1. Load dataset
# splits: tfds returns the dataset and info. tf_flowers has no standard split, so we create one.
(train_ds, val_ds), ds_info = tfds.load(
    'tf_flowers',
    split=['train[:85%]', 'train[85%:]'],
    with_info=True, as_supervised=True)

num_classes = ds_info.features['label'].num_classes
print('Number of classes:', num_classes)
print(ds_info)

# 2. Preprocessing functions
def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = tf.cast(image, tf.float32) / 255.0
    return image, label

# Data augmentation
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1),
])

def augment(image, label):
    image = data_augmentation(image)
    return image, label

# 3. Prepare datasets
train_ds = (train_ds
            .map(preprocess, num_parallel_calls=AUTOTUNE)
            .map(augment, num_parallel_calls=AUTOTUNE)
            .shuffle(1000)
            .batch(BATCH_SIZE)
            .prefetch(AUTOTUNE))

val_ds = (val_ds
          .map(preprocess, num_parallel_calls=AUTOTUNE)
          .batch(BATCH_SIZE)
          .prefetch(AUTOTUNE))

# 4. Model (simple CNN; for better results try transfer learning)
model = models.Sequential([
    layers.Input(shape=(IMG_SIZE, IMG_SIZE, 3)),
    layers.Conv2D(32, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(),
    layers.Flatten(),
    layers.Dropout(0.5),
    layers.Dense(128, activation='relu'),
    layers.Dense(num_classes, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.summary()

# 5. Train
history = model.fit(train_ds,
                    validation_data=val_ds,
                    epochs=EPOCHS)

# 6. Save model
os.makedirs(MODEL_DIR, exist_ok=True)
model.save(os.path.join(MODEL_DIR, 'flowers_cnn.keras'))

# 7. Plot training history
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(len(acc))

plt.figure(figsize=(12,4))
plt.subplot(1,2,1)
plt.plot(epochs_range, acc, label='train acc')
plt.plot(epochs_range, val_acc, label='val acc')
plt.legend()
plt.title('Accuracy')

plt.subplot(1,2,2)
plt.plot(epochs_range, loss, label='train loss')
plt.plot(epochs_range, val_loss, label='val loss')
plt.legend()
plt.title('Loss')
plt.show()

# 8. Single image prediction example
import numpy as np
import tensorflow_datasets as tfds

# load label names from ds_info
labels = ds_info.features['label'].names

# get one sample from validation set for demonstration
val_iter = iter(val_ds.unbatch().batch(1))
sample_image, sample_label = next(val_iter)
pred_probs = model.predict(sample_image)
pred_class = np.argmax(pred_probs, axis=1)[0]
print('True label:', labels[int(sample_label.numpy()[0])])
print('Predicted label:', labels[pred_class])
```
